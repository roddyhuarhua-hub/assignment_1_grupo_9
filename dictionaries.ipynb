En esta sección se trabaja con un diccionario que almacena la población de tres
ciudades del Perú. Un diccionario guarda los datos como pares `clave: valor`,
donde la clave (el nombre de la ciudad) permite acceder directamente a su valor
(la población).

## 1. Crear el diccionario e imprimirlo

Se crea el diccionario `population` con las tres ciudades solicitadas.

```python
population = {
    "Lima": 10092000,
    "Arequipa": 1008000,
    "Trujillo": 919000
}

print("Diccionario completo:", population)
```

## 2. Mostrar las llaves con `keys()`

El método `keys()` devuelve las claves del diccionario, es decir, los nombres
de las ciudades.

```python
print("Llaves del diccionario:", population.keys())
```

## 3. Obtener la población de "Lima" con `get()`

`get()` busca el valor asociado a una clave. A diferencia de `population["Lima"]`,
si la clave no existe devuelve `None` en lugar de producir un error.

```python
print("Población de Lima:", population.get("Lima"))
```

## 4. Agregar "Cusco" al diccionario

Para agregar un nuevo par `clave: valor` basta con asignarlo directamente.

```python
population["Cusco"] = 428000

print("Después de agregar Cusco:", population)
```

## 5. Eliminar "Trujillo" con `pop()`

`pop()` elimina la clave indicada y además devuelve el valor que tenía, por lo
que se puede guardar antes de perderlo.

```python
eliminado = population.pop("Trujillo")

print("Población eliminada (Trujillo):", eliminado)
```

## 6. Imprimir el diccionario final

El diccionario final ya no contiene a Trujillo y sí incluye a Cusco.

```python
print("Diccionario final:", population)
print("Número de ciudades:", len(population))
```

## Conclusión

Los diccionarios permiten acceder a un dato por su nombre y no por su posición,
como ocurre en las listas. En este ejercicio se usaron las operaciones básicas:
consultar con `get()`, listar claves con `keys()`, agregar mediante asignación
directa y eliminar con `pop()`.
