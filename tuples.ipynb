Una tupla es una colección **ordenada e inmutable**. Se parece a una lista en
que mantiene el orden y se accede por índice, pero una vez creada su contenido
no puede cambiarse.

## 1. Crear la tupla e imprimirla

Se define la tupla `ages` usando paréntesis, con las cinco edades del enunciado.

````python
ages = (20, 25, 22, 30, 28)

print("Tupla original:", ages)
````

## 2. Primer y último elemento

Se accede por posición: el índice `0` es el primer elemento y el índice `-1`
permite llegar al último sin conocer el tamaño de la tupla.

````python
print("Primer elemento:", ages[0])
print("Último elemento:", ages[-1])
````

## 3. Valor máximo, mínimo y número de elementos

Aunque una tupla no se puede modificar, sí se puede **leer**, por lo que
`max()`, `min()` y `len()` funcionan igual que en una lista.

````python
print("Valor máximo:", max(ages))
print("Valor mínimo:", min(ages))
print("Número de elementos:", len(ages))
````

## 4. Intentar modificar un elemento

Se intenta reasignar `ages[0]`. La línea está comentada para que el notebook
pueda ejecutarse completo sin detenerse, tal como permite el enunciado.

````python
# Intento de modificación:
# ages[0] = 21

# Las tuplas son inmutables en Python.
# Esto significa que sus elementos no pueden modificarse después de crear la tupla.
# Por eso, intentar cambiar ages[0] produce un TypeError.
````

## ¿Por qué Python no permite modificar una tupla?

Las tuplas son **inmutables**: al crearse, Python reserva su contenido de forma
fija y el tipo `tuple` no implementa la operación de asignación por índice
(`__setitem__`). Por eso una línea como `ages[0] = 21` lanza:

```
TypeError: 'tuple' object does not support item assignment
```

Esta inmutabilidad es intencional y resulta útil porque:

- **Protege los datos** que no deberían cambiar durante la ejecución.
- **Permite usarlas como claves de un diccionario**, ya que su valor no varía
  (las listas, al ser mutables, no pueden usarse como claves).

Si se necesita un valor distinto, no se modifica la tupla: se crea una nueva,
por ejemplo `ages = (21,) + ages[1:]`.
