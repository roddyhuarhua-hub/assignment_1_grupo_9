Una tupla es una colección ordenada e **inmutable**: una vez creada, su
contenido no puede cambiarse.

## 1. Crear la tupla y mostrarla

````python
ages = (20, 25, 22, 30, 28)

ages
````

## 2. Primer y último elemento

El índice `-1` llega al último elemento sin necesidad de conocer el tamaño.

````python
extremos = {
    "Primer elemento": ages[0],
    "Último elemento": ages[-1],
}

for etiqueta, valor in extremos.items():
    print(f"{etiqueta}: {valor}")
````

## 3. Valor máximo, mínimo y número de elementos

````python
resumen = {
    "Valor máximo": max(ages),
    "Valor mínimo": min(ages),
    "Número de elementos": len(ages),
}

for etiqueta, valor in resumen.items():
    print(f"{etiqueta}: {valor}")
````

## ¿Por qué Python no permite modificarla?

Las tuplas son inmutables: el tipo `tuple` no implementa la asignación por
índice (`__setitem__`), así que `ages[0] = 21` lanza

```
TypeError: 'tuple' object does not support item assignment
```

Esto protege los datos que no deben cambiar y permite usar tuplas como claves
de un diccionario. Si se necesita otro valor, se crea una tupla nueva:
`ages = (21,) + ages[1:]`.
