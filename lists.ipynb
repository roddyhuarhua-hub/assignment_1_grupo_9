Una lista es una colección **ordenada y mutable**: mantiene el orden en que se
insertan los elementos y permite agregarlos, quitarlos o reordenarlos después
de haberla creado. Se escribe entre corchetes `[ ]`.

En esta sección se trabaja con una lista de notas.

## 1. Crear la lista e imprimirla

Se define `grades` con las cinco notas del enunciado y se muestra tal como fue
creada, antes de modificarla.

```python
grades = [15, 18, 12, 20, 16]

print(f"Lista original: {grades}")
```

## 2. Agregar la nota 17 con `append()`

`append()` agrega un elemento **al final** de la lista. Modifica la lista
original y devuelve `None`, por eso se llama como `grades.append(17)` y no como
`grades = grades.append(17)` (esto último dejaría `grades` en `None`).

```python
grades.append(17)

print(f"Después de agregar el 17: {grades}")
```

## 3. Ordenar las notas con `sort()`

`sort()` ordena la lista de menor a mayor modificándola directamente, por lo
que tampoco se asigna a una variable nueva.

> Si se necesitara conservar el orden original, se usaría `sorted(grades)`, que
> devuelve una lista nueva y deja intacta la original.

```python
grades.sort()

print(f"Lista ordenada de menor a mayor: {grades}")
```

## 4. Nota máxima, nota mínima y total de notas

Se usan `max()`, `min()` y `len()`. Los resultados se guardan en variables en
lugar de calcularlos dentro de cada `print`: así cada función se ejecuta una
sola vez y el valor queda disponible por si se necesita más adelante.

```python
nota_maxima = max(grades)
nota_minima = min(grades)
total_notas = len(grades)

print(f"Nota más alta : {nota_maxima}")
print(f"Nota más baja : {nota_minima}")
print(f"Total de notas: {total_notas}")
```

## 5. Imprimir la lista final

Resultado después de agregar el 17 y ordenar los elementos.

```python
print(f"Lista final: {grades}")
```

## Conclusión

Las listas son **mutables**: métodos como `append()` y `sort()` modifican la
lista original en lugar de devolver una copia. Esa es la diferencia principal
frente a las tuplas, que se revisan en la Parte 2.

Un detalle de eficiencia: como la lista ya quedó ordenada, la nota mínima y la
máxima también se podrían leer directamente con `grades[0]` y `grades[-1]`, en
un solo acceso. Aquí se usaron `min()` y `max()` porque son las funciones que
pide el enunciado y porque funcionan aunque la lista no esté ordenada.
