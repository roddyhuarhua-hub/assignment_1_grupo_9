```python
grades = [15, 18, 12, 20, 16]

print("Lista original:", grades)
```

```python
grades.append(17)

print("Lista después de agregar 17:", grades)
```

```python
grades.sort()

print("Lista ordenada:", grades)
```

```python
print("Cantidad de notas:", len(grades))
print("Nota mínima:", min(grades))
print("Nota máxima:", max(grades))
```

```python
print("Lista final:", grades)
```
