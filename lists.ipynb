Una lista es una colección ordenada y **mutable**: se puede modificar después
de haberla creado.

## 1. Crear la lista y mostrarla

```python
grades = [15, 18, 12, 20, 16]

grades
```

## 2. Agregar la nota 17 con `append()`

`append()` agrega el elemento al final y modifica la lista original.

```python
grades.append(17)

grades
```

## 3. Ordenar con `sort()`

`sort()` ordena la misma lista de menor a mayor. `sorted()`, en cambio,
devuelve una lista nueva y deja intacta la original.

```python
grades.sort()

grades
```

## 4. Nota máxima, nota mínima y total de notas

```python
resumen = {
    "Nota más alta": max(grades),
    "Nota más baja": min(grades),
    "Total de notas": len(grades),
}

for etiqueta, valor in resumen.items():
    print(f"{etiqueta}: {valor}")
```
