NumPy es la librería base del cálculo numérico en Python. Su estructura
principal es el **array**, que a diferencia de una lista guarda todos sus
elementos del mismo tipo y en un bloque contiguo de memoria. Eso permite
aplicar una operación sobre todos los elementos a la vez (*vectorización*),
sin escribir bucles y de forma mucho más rápida.

## 1. Importar NumPy

Se importa con el alias `np`, que es la convención estándar.

```python
import numpy as np

np.__version__
```

## 2. Crear un array con los números del 1 al 10

`np.arange(inicio, fin)` genera los valores desde `inicio` hasta `fin - 1`: el
límite superior es **excluyente**, por eso se escribe `11` para llegar al 10.

```python
numeros = np.arange(1, 11)

numeros
```

## 3. Valor máximo y mínimo

Las funciones `np.max()` / `np.min()` y los métodos equivalentes
`numeros.max()` / `numeros.min()` hacen exactamente lo mismo.

```python
extremos = {
    "Valor máximo": numeros.max(),
    "Valor mínimo": numeros.min(),
}

for etiqueta, valor in extremos.items():
    print(f"{etiqueta}: {valor}")
```

## 4 y 5. Array de 5 ceros y array de 5 unos

`np.zeros()` y `np.ones()` crean arrays del tamaño indicado rellenos con ese
valor. Por defecto son de tipo `float64`, por eso se muestran como `0.` y `1.`
en lugar de `0` y `1`.

```python
ceros = np.zeros(5)
unos = np.ones(5)

arrays = {
    "Array de 5 ceros": ceros,
    "Array de 5 unos": unos,
    "Tipo de dato": ceros.dtype,
}

for etiqueta, valor in arrays.items():
    print(f"{etiqueta}: {valor}")
```

## 6 y 7. Crear la matriz y mostrarla

`np.array()` construye un array a partir de listas anidadas: cada lista interna
es una fila.

```python
matrix = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

matrix
```

## 8. Dimensiones de la matriz con `.shape`

`.shape` devuelve una tupla `(filas, columnas)`: en este caso 2 filas y 3
columnas.

```python
matrix.shape
```

## Atributos de un array

Todo array expone atributos que describen su estructura, sin necesidad de
recorrerlo:

| Atributo | Qué indica |
| --- | --- |
| `.shape` | forma: filas y columnas |
| `.ndim` | número de dimensiones (1 = vector, 2 = matriz) |
| `.size` | cantidad total de elementos |
| `.dtype` | tipo de dato de los elementos |

```python
datos = np.array([10, 20, 30, 40, 50])

atributos = {
    "Array": datos,
    "Forma": datos.shape,
    "Dimensiones": datos.ndim,
    "N° elementos": datos.size,
    "Tipo de dato": datos.dtype,
}

for etiqueta, valor in atributos.items():
    print(f"{etiqueta}: {valor}")
```

## Los mismos atributos en una matriz de 3x3

Nótese que ahora `.ndim` es 2 y `.size` es 9, es decir, filas por columnas.

```python
matriz = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

matriz
```

```python
atributos = {
    "Forma": matriz.shape,
    "Dimensiones": matriz.ndim,
    "N° elementos": matriz.size,
}

for etiqueta, valor in atributos.items():
    print(f"{etiqueta}: {valor}")
```

## Indexación de filas y columnas

En una matriz se indica `[fila, columna]`. Los dos puntos `:` significan
"todos", por lo que `matriz[0, :]` es la primera fila completa y `matriz[:, 1]`
es la segunda columna completa.

```python
seleccion = {
    "Elemento en fila 1, columna 1": matriz[1, 1],
    "Primera fila": matriz[0, :],
    "Segunda columna": matriz[:, 1],
}

for etiqueta, valor in seleccion.items():
    print(f"{etiqueta}: {valor}")
```

## Operaciones estadísticas

Estas funciones recorren toda la matriz y devuelven un solo valor. No hace
falta ningún bucle: NumPy opera sobre el array completo.

```python
estadisticas = {
    "Suma total": matriz.sum(),
    "Promedio": matriz.mean(),
    "Valor máximo": matriz.max(),
    "Valor mínimo": matriz.min(),
}

for etiqueta, valor in estadisticas.items():
    print(f"{etiqueta}: {valor}")
```

## Promedios por eje

El parámetro `axis` define la dirección del cálculo:

- `axis=0` recorre **las filas** hacia abajo, así que devuelve un promedio
  **por columna**.
- `axis=1` recorre **las columnas** hacia el costado, así que devuelve un
  promedio **por fila**.

```python
promedios = {
    "Promedio por columna (axis=0)": matriz.mean(axis=0),
    "Promedio por fila    (axis=1)": matriz.mean(axis=1),
}

for etiqueta, valor in promedios.items():
    print(f"{etiqueta}: {valor}")
```

## Conclusión

NumPy permite aplicar operaciones sobre todos los elementos de un array a la
vez, sin escribir bucles. Los atributos `.shape`, `.ndim`, `.size` y `.dtype`
describen la estructura del array, mientras que funciones como `sum()`,
`mean()`, `max()` y `min()` resumen su contenido. El parámetro `axis` permite
además calcular esos resúmenes por fila o por columna.
