```python
import numpy as np

print("NumPy funciona correctamente")
print("Versión:", np.__version__)
```

```python
numeros = np.arange(1, 11)

print("Array del 1 al 10:", numeros)
```

```python
print("Valor máximo:", np.max(numeros))
print("Valor mínimo:", np.min(numeros))
```

```python
ceros = np.zeros(5)
unos = np.ones(5)

print("Array de 5 ceros:", ceros)
print("Array de 5 unos:", unos)
```

```python
matrix = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print("Matriz:")
print(matrix)

print("Dimensiones de la matriz:", matrix.shape)
```

```python
datos = np.array([10, 20, 30, 40, 50])

print(datos)
```

```python
print("Cantidad de elementos:", datos.size)
print("Dimensiones:", datos.ndim)
print("Tipo de dato:", datos.dtype)
```

```python
matriz = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print(matriz)
```

```python
print("Forma de la matriz:", matriz.shape)
print("Dimensiones:", matriz.ndim)
print("Cantidad de elementos:", matriz.size)
```

```python
print("Elemento fila 1, columna 1:", matriz[1, 1])
print("Primera fila:", matriz[0, :])
print("Segunda columna:", matriz[:, 1])
```

```python
print("Suma total:", np.sum(matriz))
print("Promedio:", np.mean(matriz))
print("Valor máximo:", np.max(matriz))
print("Valor mínimo:", np.min(matriz))
```

```python
print("Promedio por columna:", np.mean(matriz, axis=0))
print("Promedio por fila:", np.mean(matriz, axis=1))
```
